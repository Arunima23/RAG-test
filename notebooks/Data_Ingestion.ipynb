```python
from langchain_community.document_loaders import DirectoryLoader, TextLoader

# Initialize DirectoryLoader with configuration arguments
dir_loader = DirectoryLoader(
    path="data/text_files",
    glob="**/*.txt",
    loader_cls=TextLoader,
    loader_kwargs={"encoding": "utf-8"},
    show_progress=False
)

# Execute .load() without arguments
documents = dir_loader.load()
documents
```

```python
import logging
from langchain_community.document_loaders import DirectoryLoader, PyMuPDFLoader

# Suppress logging clutter
logging.getLogger("pymupdf").setLevel(logging.ERROR)

# Point to your subfolder containing PDFs
dir_loader = DirectoryLoader(
    path="Input Files/pdf_files",
    glob="*.pdf",
    loader_cls=PyMuPDFLoader,
    show_progress=False
)

pdf_documents = dir_loader.load()

print(f"Loaded total {len(pdf_documents)} pages across all PDF files.\n")

if pdf_documents:
    for i, doc in enumerate(pdf_documents, 1):
        print(f"=== PAGE {i} METADATA ===")
        print(doc.metadata)
        print("\n=== PAGE CONTENT (FIRST 300 CHARS) ===")
        # Replacing newlines with spaces for clean display
        clean_text = doc.page_content[:300].replace('\n', ' ')
        print(f"{clean_text}...")
        print("-" * 60 + "\n")
else:
    print("No PDF files found.")
```

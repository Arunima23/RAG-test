```python
import chromadb
from sentence_transformers import SentenceTransformer
```

```python
# Connect to existing persistent database
client = chromadb.PersistentClient(
    path=r"E:\RAG\RAG-test\data\vector_store"
)

# Access existing collection
collection = client.get_collection(
    name="pdf_documents"
)

print("Connected successfully!")
print("Total stored chunks:", collection.count())
```

```python
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
query = "Areal data on a spatial graph"

# Generate query embedding
query_embedding = embedding_model.encode([query])

# Search ChromaDB
results = collection.query(
    query_embeddings=query_embedding.tolist(),
    n_results=3,
    include=["documents", "metadatas", "distances"]
)
```

```python
for i in range(len(results["documents"][0])):

    print("\n" + "=" * 70)

    print(f"RESULT {i + 1}")

    print("\nRETRIEVED CONTENT:")
    print(results["documents"][0][i])

    print("\nSOURCE METADATA:")
    print(results["metadatas"][0][i])

    print("\nDISTANCE:")
    print(results["distances"][0][i])
```

```python
queries = [
    "How are observations transformed using a reference value and scale?",

    "What is an induced subgraph and how are connected components defined?",

    "How does Betti-0 describe spatial connectivity across thresholds?"
]

for query in queries:

    print("\nQUERY:", query)

    query_embedding = embedding_model.encode([query])

    results = collection.query(
        query_embeddings=query_embedding.tolist(),
        n_results=3,
        include=["documents", "metadatas", "distances"]
    )

    for i, content in enumerate(results["documents"][0]):
        print(f"\nResult {i+1}:")
        print(content)
        print("Metadata:", results["metadatas"][0][i])
        print("Distance:", results["distances"][0][i])
```

```python
for i, (content, metadata, distance) in enumerate(
    zip(
        results["documents"][0],
        results["metadatas"][0],
        results["distances"][0]
    ),
    start=1
):

    print(f"\n{'=' * 70}")
    print(f"RANK: {i}")
    print(f"DISTANCE: {distance:.6f}")

    print("\nSOURCE:")
    print(metadata.get("source_file"))

    print("PAGE:")
    print(metadata.get("page_label", metadata.get("page")))

    print("\nCONTENT:")
    print(content)
```

```python
from collections import Counter

# Fetch all stored documents
all_data = collection.get(
    include=["documents", "metadatas"]
)

documents = all_data["documents"]
ids = all_data["ids"]
metadatas = all_data["metadatas"]

# Count exact duplicate content
content_counts = Counter(documents)

duplicates = {
    content: count
    for content, count in content_counts.items()
    if count > 1
}

print("Total stored chunks:", len(documents))
print("Unique content chunks:", len(content_counts))
print("Duplicate content groups:", len(duplicates))
print(
    "Extra duplicate entries:",
    sum(count - 1 for count in duplicates.values())
)
```

```python
from collections import defaultdict

content_map = defaultdict(list)

all_data = collection.get(
    include=["documents", "metadatas"]
)

for doc_id, document, metadata in zip(
    all_data["ids"],
    all_data["documents"],
    all_data["metadatas"]
):
    content_map[document].append({
        "id": doc_id,
        "source": metadata.get("source_file"),
        "page": metadata.get("page_label"),
        "doc_index": metadata.get("doc_index")
    })

# Display five duplicate groups
shown = 0

for content, records in content_map.items():
    if len(records) > 1:
        print("\n" + "=" * 80)
        print("Duplicate count:", len(records))
        print("Content preview:", content[:250])
        print("Records:")

        for record in records[:5]:
            print(record)

        shown += 1

    if shown == 5:
        break
```
